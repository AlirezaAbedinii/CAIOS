# Object detection, three ways

This notebook runs inside the **YOLO module** from the CAIOS marketplace: the
model's own code and environment, opened as a workspace instead of an API.

```python
from ultralytics import YOLO
from ultralytics.utils import ASSETS
from PIL import Image

model = YOLO("yolov8n.pt")                          # the model this module serves
result = model(ASSETS / "bus.jpg", verbose=False)[0]
Image.fromarray(result.plot()[..., ::-1])            # its detections, drawn
```

```python
from collections import Counter

found = Counter(result.names[int(c)] for c in result.boxes.cls)
found
```

## The same model, as a serverless service

```python
import caios_demo as caios

remote = caios.serverless(ASSETS / "bus.jpg")        # the low-code tier's endpoint
Counter(d["name"] for d in remote)
```

## And the private language model

```python
caios.ask(f"Describe, in one plain sentence, a street photo where YOLO detected {dict(found)}.")
```
